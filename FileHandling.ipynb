```python
# Navigating the file: tell()
```

```python
# file_object.tell() : acts as the program's GPS 

# Returns an integer specifying the exact byte position from the beginning of the file to the current location of the file object
```

```python
import os 

os.chdir(r"F:\python practice") 
with open("sample.txt","rb") as f:
    print( f.tell()) # will tell the current index of the pointer 
    print(f.read(10))

    f.seek(3,0) # second parameter as 2 will work in binary mode only 

    f.seek(-10,2) # wouldnt work as file is not in binary mode 
    print( f.tell())
    # print(f.read())
    print(f.read().decode('utf-8')) # it will decode the returned byte into string 
    
```
