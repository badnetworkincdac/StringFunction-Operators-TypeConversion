# String Functions 

```python
# the string is immutable in nature in python so functions applied does not change the original string , it will create copy of it with the change
```

```python
st = "this is string example " # basic python program
```

```python
st.capitalize()
```

```python
# count returns the number of occurence of the sub string in the given string 

sub="is"

st.count(sub)
```

```python
st.upper()
```

```python
st.lower()
```

```python
st.count("is",1,10)
```

```python
# find function is used to get the index number(first location) of character in the string 

st.find("n")
```

```python
st.find("st")
```

```python
# index function is similar to find , only difference is that if index number is not existing then index will give error and find will give -1
st.index("str")
```

```python
# Swapcase function will swap the upper and lower case of the string to the opposite 
st.swapcase()
```

```python
# title function will convert the first character of each word in a string to upper case
st.title()
```

```python
# Replace function is used to replace the old value with the new value 
# replace(old value, new value, number of occurence to replace) 
st.replace("x","z",1) # here only 1st occurence of x will be replaced by z 
st.replace("i","d",2) # here only first 2 occurence of i will be replaced by d
```

```python
# len function will return the length of the string , it is not string function 
len(st)
```

# Conversion Functions 

```python
# int conversion function 
a='1010'
d=int(a)
print(d) 

e = int(a,2) # changing its default from decimal to binary 
print(e)

e = int(a,8) # changing its default from decimal to octal 
print(e)
```

```python
h= 'AF' 
d= int(h,16) # changing its default from decimal to hexadecimal 
print(d)
```

```python
bin(175) # binary representation of a number 
```

```python
# Prefix for various number system 

# Ob -> binary 
# Oo -> octal 
# Ox -> hexadecimal
```

```python
hex(175)
```

```python
oct(1010)
```

```python
chr(23) # convert the interger to a character based on the ascii number 
```

```python
ord('\x17') # convert character to a integer value 
```

```python
chr(65)
```

```python
ord('c') # works only with single character only 
```
