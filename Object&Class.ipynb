```python
class circle:
    Pi=1.4  # class variable
    def m1(self):
        print("This is a sample method")
    def Area(self,radius):
        self.R1 = radius # instance variable 
        print("Area of Circle : ",(circle.Pi*(self.R1**2)))
```

```python
 # constructor 
class addition:
    def __init__(self):
        self.FN=10
        self.SN=20 

    def add(self):
        print("Result of addition : ",(self.FN+self.SN)) 

    def __del__(self):
        print("Destructor is called")

 # constructor 
class addition1:
    def __init__(self,fn,sn):
        self.FN=fn
        self.SN=sn 

    def add(self):
        print("Result of addition : ",(self.FN+self.SN)) 

a = addition()
a.add()
```

```python
b= addition1(1,2) 
b.add()
```
