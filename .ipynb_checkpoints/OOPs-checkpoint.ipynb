# Inheritance 

```python
# Heirarichal inheritance 

class vehicle:
    def vehicle_info(self):
        print("Vechicle class") 

class car(vehicle):
    def car_info(self):
        print("Car class")


class truck(vehicle):
    def truck_info(self):
        print("Truck class")

c = car() 
c.car_info()
c.vehicle_info()

t = truck() 
t.truck_info()
t.vehicle_info()
```

```python
# Multilevel 

class sportscar(car):
    def sports(self):
        print("Sports car")

s = sportscar() 
s.sports()
s.car_info()
s.vehicle_info()
```

```python
# Multiple inheritance 

class van(car,truck):
    def van_info(self):
        print("Van class") 

v = van()
v.van_info()
v.car_info() 
v.truck_info() 
v.vehicle_info()
```

```python
# super keyword/function 

class company:
    def compname(self):
        return "Cdac noida" 

class emp(company):
    def emp_info(self):
        v=super().compname()
        print("emp works for ,",v)

e = emp()
e.emp_info()
```

```python
# for diamond problem in hybrid inheritance we use MRO (Method Resolution Order - Depth first left right) 
# it will consider left side class written in the definition of inherited class 
```

```python
class p1:
    pass 

class c1: 
    def m1(self):
        print("C1 class")

class c2: 
    def m1(self):
        print("C2 class")

class c12(c2,c1): # here it will use c2 class as it is on left side of its definiton
    pass 

    
c12.mro()

print(c12.mro()) 

obj = c12() 
obj.m1()

# to access other class function c1
c1.m1(obj)
```

```python
# Constructor chaining 

class vehicle:
    def __init__(self,engine):
        print("Inside vehicle constructor")
        self.engine = engine 

class car(vehicle):
    def __init__(self,engine,max_speed):
        super().__init__(engine) 
        print("Inside car constructor")
        self.max_speed = max_speed 

class electric_car(car):
    def __init__(self,engine,max_speed,km_range):
        super().__init__(engine,max_speed)
        print("Inside electric car constructor") 
        self.km_range= km_range

ev = electric_car("2000cc",240,700)
print(f" Engine={ev.engine} , Max Speed = {ev.max_speed} and km range = {ev.km_range}") 

print(electric_car.mro())
```

# Polymorphism

```python
# One thing many forms 

# compile time polymorphism: 
# Overloading: 
#     Method Overloading 
#     Operator Overloading 

# Runtime Polymorphism: 
#     method overriding 
```

```python
class vehicle:
    def max_speed(self):
        print("Max speed is 120") 

class car(vehicle):
    def max_speed(self):
        print("max speed is 200") 

c = car()
c.max_speed()
```

```python
# Function overloading is not directly possible in python due to its dynamic nature but we use it using decorators 
```

```python
# Operator overloading is possible 
```

```python
class Point:
    def __init__(self,x=0,y=0):
        self.x=x 
        self.y=y 

    def __str__(self):
        # return "({0},{1})".format(self.x,self.y) 
        return "{0}+j{1}".format(self.x,self.y) 

    def __add__(self,other):
        x=self.x+other.x 
        y=self.y+other.y 
        return Point(x,y) 

p1 =Point(4,5)
print(p1)

p2=Point(2,3)
print(p2) 

print(p1+p2) 
```

# Abstraction 

```python
# we use Abstract class to implement abstraction 
# abstract class has abstract method and non abstract method 
```

```python
# from abc module import ABC(Abstract Base Class) : every user defined abstract class need to inherit this ABC class 
```

```python
from abc import ABC, abstractmethod 
```

```python
class ac1(ABC):
    @abstractmethod 
    def some(self):
        print("Implementation in Parent Class") 

    def m1(self):
        print("This is non abstract method")

class anotherclass(ac1):

    def m2(self):
        print("This is non abstract method of another class")

    # if we dont redefine some method here it will not work 
    def some(self):
        print("Redefining in Child class")
        super().some()

        
a = anotherclass() 
a.m2()
a.m1()
a.some()
```
