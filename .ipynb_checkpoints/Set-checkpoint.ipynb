# Set operations 

```python
a={1,2,3,4,5}
b={5,6,7,8}

# union operator 
a|b
```

```python
a.union(b)
```

```python
# intersection operator 
a&b
```

```python
a.intersection(b)
```

```python
b.intersection(a)
```

```python
# difference operator 
a-b
```

```python
b-a
```

```python
a.difference(b)
```

```python
b.difference(a)
```

```python
# Symmetric difference 
a^b
```

```python
b^a
```

```python
a.symmetric_difference(b)
```

```python
b.symmetric_difference(a)
```

# Python Frozenset 

```python
# immutable set 
# read only set
# cannot add or update values in frozenset 
```

```python
a = frozenset(1,2,3,4,5)
b= frozenset(4,5,6,7,8)
```

# dynamic input from user 

```python
# in list 
l1 = input("enter values : ").split(",")
```

```python
l1
```

```python
l2 = []
for i in l1:
    l2.append(int())
print(l2)
```

```python
# list comprehension used to take input from user and convert it into list dynamically
```

```python
x=[i for i in range(10)]
x
```

```python
y = [2 for i in range(10)]
y
```

```python
x=[i for i in range(10) if i%2==0 ]
x
```

```python
# dynamic input 
z =[int(i) for i in input("enter value: ").split(",")]
z
```

```python
# taking fixed input from user 
k = [ input("enter value "+str(i+1)+" : ") for i in range(5)]
k
```
