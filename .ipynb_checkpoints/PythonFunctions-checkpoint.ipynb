```python
def add(a,b):
    """ This function is used for addition of two numbers """
    result=a+b
    print("Result:",result) 

add(2,3)
```

```python
# to access the docstring used inside the function 
add.__doc__
```

```python
# non default parameter 

def greet(msg):
    print("Hi ,",msg)

greet("Hello world")
```

```python
# default arguments 
# default arguments should be placed after the non default arguments or else it will give error
def greet(name,msg=" Namaste Ji"):
    print("Hi ,"+name+msg)

greet("Hello world")
```

```python
# Keyword arguments
def greet(name,msg):
    print("hi, "+name+msg)

greet("hello", msg=" how are u")
# greet(name="hello"," how are u")  : this is not keyword arguments
```

```python
# *args argument used to passed multiple arguments 
# it will take parameters passed in the function call as a tuple 

def add2(*num):
    s=0 
    for i in num:
        s=s+i 
    print(s)

add2(1,2,3,4,5)
```

```python
# **kargs argument 
# it will take **num as dictionary and we can pass any number of keyword arguments

def add3(**num):
    print(num)

add3(x="cdac",y=" noida",z=" cdot")
```

```python
# Python anonymous/lambda function 

# lambda parameter(s):expression 

double = lambda x:x*2 ## double variable will become function 

double(2)
```

```python
# Map and filter in lambda 

def calsquare(n):
    return n*n 

lt=[1,2,3,4,5] 

lt=list(map(calsquare,lt)) # map will return address of the map so we will convert it into list
lt
```

```python
l1 = ["cdac","cdot","nic","drdo"] 
# to convert it into uppercase 
l1 = list(map(str.upper,l1))
print(tuple(map(str.upper,l1)))
l1
```

```python
lt=list(map(lambda x:x**2,lt))
lt
```

```python
lt = list(filter(lambda x:x<100,lt))
lt
```
