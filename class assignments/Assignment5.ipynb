```python
# 3. Write a program to display powers of 2 using Anonymous function 

n = int(input("Enter the total numbers you want to enter : ")) 

lt = [int(input("Enter number :")) for i in range(n) ]

print("User Entered List : ",lt) 

# lt = [ (lambda x : x**2) (i) for i in lt ]
lt = list(map(lambda x : x**2,lt))

print("Modified list in power of 2 : ",lt)
```

```python
# 4. WAP to find the numbers divisible by thirteen from the list using anonymous function 

n = int(input("Enter the total numbers you want to enter : ")) 

lt = [int(input("Enter number :")) for i in range(n) ]

print("User Entered List : ",lt) 

lt = list(filter(lambda x : x%13==0,lt))

print("Modified list  : ",lt)
```

```python
# 5. WAP to display the fibonacci series using recursive function 

def fib(n: int):
    if n == 0:
        return 0
    elif n == 1:
        return 1
    else:
        return fib(n - 1) + fib(n - 2)

 

n = int(input("Enter number : ")) 

for i in range(n):
    print(fib(i),end=" ")
```

```python
# 6. WAP to find the sum of natural numbers up to n using recursive function

def total(n:int):
    t = 0 
    for i in range(1,n+1):
        t+=i 
    return t

n = int(input("Enter number :")) 

print(f"Sum of number upto {n} : {total(n)}")
```

```python
# 7. Palindrome recognizer  

def palindrome_recognizer(st:str):
    st = st.lower()
# print(st)
    st2=list(st)
    lt = list(filter(lambda s : s.isalnum() and s!=" ",st2))

    if lt!= lt[::-1]:
        return False
    else:
        return True



st = input("Enter a string : ") 

if palindrome_recognizer(st):
    print("String is palindrome and accepted")
else: 
    print("String entered is not palindrome so not accepted")
```

```python
# 8. Check pangram 
def check_pangram(st:str):
    st = st.lower()
    st2=list(st)
    lt = list(filter(lambda s : s.isalnum() and s!=" ",st2))
    unique={}

    for i in lt:
        unique[i]=unique.get(i,0)+1 

        
    return len(unique.keys())==26

st = input("Enter a string : ") 


if check_pangram(st):
    print("Entered string is Pangram")
else: 
    print("Entered string is not a Pangram")
```

```python
# 9. Define a function overlapping() that takes two lists and returns True if they have at least one member in common , False otherwise 

def overlapping(list1=None,list2=None):

    for item in list1:
        if item in list2:
            return True;

    return False

n = int(input("Enter the number of elements you want in list1 : "))

m = int(input("Enter the number of elements you want in list2 : "))

list1=[int(input("Enter value")) for i in range(n) ] 

list2 = [int(input("Enter value: ")) for i in range(m)]

if overlapping(list1,list2):
    print("Lists have some common or overlapping values ")
else:
    print("Lists does have any common or overlapping values ")
```

```python
# 10. Write a function find_longest_word() that takes a list of words and returns the lenght of the longest one 

def find_longest_word(st:str):
    lt = list(st.split())
    maxlength=0
    

    for i in lt:
        curlength = len(i)
        if curlength>maxlength:
            maxlength=curlength
            temp = i

    return temp 


st = input("Enter the string : ") 

ans = find_longest_word(st)

print("Longest word in a string : ",ans)
```

```python
# 11. Write a function filter_long_words() that takes a list of words and an integer n and returns the list of words that are longer than n 

def filter_long_words(lt=None, n = int):
    res = []

    for i in lt:
        if len(i) > n:
            res.append(i)

    return res

st =input("Enter the string : ")  

n = int(input("Enter the integer n : ")) 

lt = list(st.split()) 

ans = filter_long_words(lt,n) 

print("Result :",ans)
```
