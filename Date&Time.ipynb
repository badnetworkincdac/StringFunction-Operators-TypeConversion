```python
# to customize date and time format we use , strftime
```

```python
from datetime import datetime 
x=datetime.now()

print("Current date and time: ",x)
print(x.strftime("%a, %d %B %Y"))
d = datetime(2026,10,24,12,15,30)
print(d)
```

```python
print(d.strftime("One %d-%m-%y at %I:%M:%S, i will be at home")) # to display time in 12 hour format 
```

```python
print(d.strftime("One %d-%m-%y at %H:%M:%S, i will be at home")) # to display time in 24 hour format 
```

```python
d1 = "2025/02/12 08:15:45"
format1 = "%Y/%m/%d %H:%M:%S"
result= datetime.strptime(d1,format1)
print(result)
```

```python
# for time difference  

# we use timedelta 
```

```python
from datetime import datetime,timedelta 
x = datetime.now()
print("Current date: ",x) 
print("Date of tomorrrow : ",x+timedelta(days=1)) 
print("Date 10 Days before today :",x-timedelta(days=10)) 
print("Date :",x+timedelta(weeks=49,days=1,hours=10,minutes=23,seconds=12))
```

```python
from datetime import datetime 
import pytz 
local = datetime.now()
print("Local Indian Time: ",local.strftime("%d-%m-%Y, %H:%M:%S")) 
tzNY = pytz.timezone("America/New_York") 
x = datetime.now(tzNY) 
print("New York Time: " ,x.strftime("%d-%m-%Y, %H:%M:%S"))
print(x) 
tzLN =  pytz.timezone("Europe/London") 
y = datetime.now(tzLN) 
print("London Time: " ,y.strftime("%d-%m-%Y, %H:%M:%S"))
print(y)
```
